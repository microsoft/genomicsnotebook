## Enhancing Genomics Annotation with GraphRAG

The intersection of generative AI and genomics is rapidly reshaping how researchers understand and annotate complex biological data. Among the emerging techniques, GraphRAG (Graph Retrieval-Augmented Generation) stands out by integrating structured knowledge graphs with large language models to enhance contextual reasoning and data retrieval. In genomics annotation, where relationships between genes, proteins, and phenotypes are intricate and deeply interlinked, GraphRAG offers a novel approach to navigate this complexity with greater precision and interpretability. This notebook is a follow up study of our previous [RAG research paper](https://doi.org/10.1093/bioadv/vbaf019) and explores how GraphRAG can be leveraged to accelerate and improve the annotation of genomic sequences, bringing AI one step closer to becoming a trusted tool in molecular biology.

### 3. Set Up Your Workspace Variables

```python
!graphrag init --root ./genomicsragtest
```

### 4. Running the Indexing pipeline

```python
!graphrag index --root ."/genomicsragtest"
```

### 5. Running the Genomics Annotation Query Engine

```python
!graphrag query \
--root ./genomicsragtest \
--method local \
--query "Annotate chr1:930220"
```

### 6. Visualize and save the graph

Attention: Please be sure that you installed the relevant libraries.

```python
import networkx as nx
import matplotlib.pyplot as plt
# Load GraphML file
G = nx.read_graphml("./genomicsragtest/output/graph.graphml")
pos = nx.spring_layout(G)
# Draw the graph
plt.figure(figsize=(50, 30))
nx.draw(G, pos, with_labels=True, node_color='skyblue', edge_color='gray', node_size=5, font_size=8)
plt.title("GraphML Visualization")
plt.savefig("graph_rag_output.png", format="png", dpi=500)
plt.show()
```
